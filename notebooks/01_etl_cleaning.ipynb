```python
# Verificando quais são valores únicos da coluna 'remote_ratio'
df['remote_ratio'].unique()

# 0 - trabalhos presenciais
# 100 - trabalhos remotos
# 50 - trabalhos híbridos
```

```python
# Mapeando os valores de 'remote_ratio' para português, pois eles estão em códigos. Isso ajudará na análise e visualização dos dados.
remote_map = {
    0: 'Presencial',
    50: 'Híbrido',
    100: 'Remoto'
}

df['remote_ratio_pt'] = df['remote_ratio'].map(remote_map)
df['remote_ratio_pt'].unique()
```

```python
# Verificando quais são valores únicos da coluna 'experience_level'
df['experience_level'].unique()
```

```python
# Mapeando os valores de 'experience_level' para português, pois os cargos estão classificados por abreviações em inglês. Isso ajudará na análise e visualização dos dados.
experience_map = {
    'EN': 'Júnior',
    'MI': 'Pleno',
    'SE': 'Sênior',
    'EX': 'Executivo'
}

df['experience_level_pt'] = df['experience_level'].map(experience_map)
df['experience_level_pt'].unique()
```

```python
# Verificando quais são valores únicos da coluna 'employment_type'
df['employment_type'].unique()
```

```python
# Mapeando os valores de 'employment_type' para português, pois os contratos estão classificados por abreviações em inglês. Isso ajudará na análise e visualização dos dados.
employment_map = {
    'FT': 'Tempo Integral',
    'PT': 'Meio Período',
    'CT': 'Contrato',
    'FL': 'Freelancer'
}

df['employment_type_pt'] = df['employment_type'].map(employment_map)
df['employment_type_pt'].unique()
```

```python
# Verificando quais são valores únicos da coluna 'company_size'
df['company_size'].unique()
```

```python
# Mapeando os valores de 'company_size' para português, pois os tamanhos estão classificados por abreviações em inglês. Isso ajudará na análise e visualização dos dados.
company_size_map = {
    'S': 'Pequena',
    'M': 'Média',
    'L': 'Grande'
}

df['company_size_pt'] = df['company_size'].map(company_size_map)
df['company_size_pt'].unique()
```

```python
# Criando uma nova coluna com o salário em milhares de dólares (K USD = mil dólares) para melhorar a visualização dos dados
df["salary_usd_k"] = df["salary_in_usd"] / 1000
```

```python
# Função para converter ISO-2 para ISO-3
def iso2_to_iso3(code):
    try:
        return pycountry.countries.get(alpha_2=code).alpha_3
    except:
        return None
```

```python
if 'employee_residence' in df.columns:
    df['residence_iso3'] = df['employee_residence'].apply(iso2_to_iso3)
```

```python
# Verificando se as novas colunas criadas foram adicionadas corretamente ao DataFrame
df.head()
```
